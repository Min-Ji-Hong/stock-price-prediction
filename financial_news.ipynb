```python
from gdeltdoc import GdeltDoc, Filters
import pandas as pd

gd = GdeltDoc()

f = Filters(
    keyword="Tesla",
    start_date="2017-01-01",
    end_date="2025-11-13"
)

# This returns a DataFrame with columns ['datetime', 'Tone']
tesla_tone = gd.timeline_search("timelinetone", f)

print(tesla_tone.head(20))
```

```python
tesla_tone.describe()
```

```python
import matplotlib.pyplot as plt

tesla_tone['datetime'] = pd.to_datetime(tesla_tone['datetime'])

plt.figure(figsize=(14,5))
plt.plot(tesla_tone['datetime'], tesla_tone['Average Tone'])
plt.title("Tesla News Tone Over Time (GDELT)")
plt.xlabel("Date")
plt.ylabel("Tone Score")
plt.grid(True)
plt.tight_layout()
plt.show()
```
